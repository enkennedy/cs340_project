```python
import matplotlib.pyplot as plt
```

Fit as number of classes approaches number of rooms * number of time slots

constants: r = 1000, t = 10, s = 1000

```python
numClasses = [10, 50, 100, 500, 1000, 5000, 10000]
avgFit1 = [1.0, 0.894, 0.885, 0.874, 0.873, 0.873, 0.872]
```

```python
plt.plot(numClasses, avgFit1)
plt.ylim(0,1)
plt.xlabel("Number of Classes")
plt.ylabel("Fit")
plt.title("Fit as Number of Classes Approaches Rooms * Time Slots")
plt.show()
```
