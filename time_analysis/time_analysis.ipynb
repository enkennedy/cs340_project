```python
import matplotlib.pyplot as plt
```

Runtime as number of students increases

```python
numStudents = [10, 20, 50, 100, 150, 200, 300, 400, 500]
times1 = [0.00016795797273516655, 0.00017445767298340797, 0.00018633389845490456, 0.000245917122811079, 0.0002954169176518917, 0.000331084243953228, 0.00044395821169018745, 0.000507791992276907, 0.0006377501413226128]

plt.plot(numStudents, times1)
plt.xlabel("Number of Students")
plt.ylabel("Runtime")
plt.title("Runtime with respect to Number of Students")
plt.show()
```
